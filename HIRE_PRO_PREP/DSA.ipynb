```python
#O(1) — constant access
arr = [10,20,30,40]
print(arr[2])
```

```python
#O(log n) — binary search
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target: return mid
        elif arr[mid] < target: lo = mid + 1
        else: hi = mid - 1
    return -1

print(binary_search([1,3,5,7,9,11], 7))
```

```python
#O(n) — linear scan
def total(arr):
    s=0
    for x in arr:
        s+=x
    return s
print(total([1,2,3,4]))
```

```python
#O(n log n) — sorting
print(sorted([5,3,6,2,4]))
```

```python
#O(n²) — nested loop, all pairs
def all_pairs(arr):
    pairs=[]
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            pairs.append((arr[i],arr[j]))
    return pairs
print(all_pairs([1,2,3,4]))
```

```python
#O(2ⁿ) — naive recursive Fibonacci
def fib_naive(n):
    if n<=1: return n
    return fib_naive(n-1) + fib_naive(n-2)
print(fib_naive(10))
```

```python
#O(n!) — permutations
from itertools import permutations
print(len(list(permutations([1,2,3,4]))))
print(permutations([1,2,3,4]))
print(list(permutations([1,2,3,4])))
```

```python
#Drop constants: two separate O(n) passes are still O(n) overall, not "2n"
def two_loops(arr):
    s1=sum(arr)
    s2=sum(X*2 for X in arr)
    return s1,s2
print(two_loops([1,2,3]))
x,y=two_loops([1,2,3])
print(x)
print(y)
```

```python
#Recursion has O(depth) space — the call stack is real memory, not "free"
def depth_count(n, current=0):
    if n==0:
        return current;
    return depth_count(n-1, current+1)
print("Max depth reached:" ,depth_count(50))
```

```python
#Strings are immutable — every += creates a brand-new object
s = "a"
id_before=id(s)
s += "b"
id_after=id(s)
print("same obejct after +=?", id_before ==id_after)
print(id_before)
print(id_after)
```

```python
#in on a list — O(n), linear scan
print(99 in [1,2,3,99, 4])
```

```python
#Slicing creates a new object — O(k) copy, not a view
a=[1,2,3]
b=a[:]
print("same object?",a is b, "| same values?", a == b)
```

```python
#dict get/set — O(1) average
d={}
d["X"]=1
print(d.get("X"), d.get("Y","default"))
```

```python
#set membership — O(1) average, vs O(n) for list
s={1,2,3}
print(2 in s, 5 in s)
```

```python
#deque — O(1) at both ends (list is only O(1) at the right end)
from collections import deque
dq=deque([2,3])
dq.append(4)
print(dq)
dq.appendleft(1)
print(dq)
p=dq.pop()
print(p)
print(dq)
l=dq.popleft()
print(l)
print(dq)
print(list(dq))
```

```python
#heapq push/pop — O(log n)
import heapq
heap=[]
for x in [5,1,3]:
    heapq.heappush(heap,x)
print(heap)
p=heapq.heappop(heap)
print(p)
print(heap)
```

```python
#bisect — O(log n) search on an already-sorted list
import bisect
sorted_list=[1,3,5,7,9]
print(bisect.bisect_left(sorted_list,5))
```
